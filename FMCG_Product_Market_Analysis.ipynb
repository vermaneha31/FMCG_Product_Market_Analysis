```python
import pandas as pd
```

```python
print("FMCG Product & Market Performance Analysis")
```

```python
# FMCG products ki basic information create kar rahe hain

products = [
    "Shampoo",
    "Face Wash",
    "Body Lotion",
    "Soap",
    "Toothpaste"
]

print(products)
```

```python
# Har product ki hamari sales units me
our_sales = [8200, 6500, 4800, 10500, 7600]

# Har product ki total market sales units me
market_sales = [12000, 10000, 8500, 14000, 11000]

print(our_sales)
print(market_sales)
```

```python
# Hamare products ki selling price
our_price = [299, 199, 249, 45, 110]

# Competitor ke similar products ki selling price
competitor_price = [279, 189, 229, 40, 99]

print(our_price)
print(competitor_price)
```

```python
# Customers ke average ratings out of 5
customer_rating = [4.2, 3.8, 3.5, 4.4, 4.0]

print(customer_rating)
```

```python
# Har product ke liye customer complaints ki main category
complaint_category = [
    "Price",
    "Quality",
    "Fragrance",
    "Availability",
    "Packaging"
]

print(complaint_category)
```

```python
# Har product ke liye primary sales region
region = [
    "North",
    "East",
    "South",
    "West",
    "North"
]

print(region)
```

```python
# Saare product data ko ek table me combine 

data = pd.DataFrame({
    "Product": products,
    "Our Sales": our_sales,
    "Market Sales": market_sales,
    "Our Price": our_price,
    "Competitor Price": competitor_price,
    "Customer Rating": customer_rating,
    "Complaint": complaint_category,
    "Region": region
})

data
```

```python
# Dataset ki first 5 rows 
data.head()
```

```python
# Graph banane ke liye Matplotlib import 
import matplotlib.pyplot as plt
```

```python
# Product-wise hamari sales aur market sales compare 

data.plot(
    x="Product",
    y=["Our Sales", "Market Sales"],
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Our Sales vs Market Sales")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Sales (Units)")

# Product names ko seedha rakh rahe hain
plt.xticks(rotation=0)

# Graph ko display kar rahe hain
plt.show()
```

```python
# Product-wise hamari price aur competitor ki price compare

data.plot(
    x="Product",
    y=["Our Price", "Competitor Price"],
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Our Price vs Competitor Price")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Price (₹)")

# Product names ko seedha rakh rahe hain
plt.xticks(rotation=0)

# Graph ko display kar rahe hain
plt.show()
```

```python
# Product-wise customer rating compare 

data.plot(
    x="Product",
    y="Customer Rating",
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Customer Rating by Product")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Rating (Out of 5)")

# Product names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Complaint categories ki frequency count 
complaint_count = data["Complaint"].value_counts()

# Complaint category ka bar graph bana rahe hain
complaint_count.plot(
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Customer Complaint Categories")

# X-axis ka naam
plt.xlabel("Complaint Category")

# Y-axis ka naam
plt.ylabel("Number of Products")

# Category names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Region-wise hamari sales ko compare 
region_sales = data.groupby("Region")["Our Sales"].sum()

# Region-wise sales ka bar graph 
region_sales.plot(
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Region-wise Sales Performance")

# X-axis ka naam
plt.xlabel("Region")

# Y-axis ka naam
plt.ylabel("Our Sales (Units)")

# Region names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Market sales aur hamari sales ke beech ka gap calculate 
data["Sales Gap"] = data["Market Sales"] - data["Our Sales"]

# Product-wise sales gap ka bar graph 
data.plot(
    x="Product",
    y="Sales Gap",
    kind="bar",
    figsize=(10, 6),
    legend=False
)

# Graph ka title
plt.title("Product-wise Sales Gap")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Sales Gap (Units)")

# Product names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Products ko hamari sales ke according highest se lowest sort 
product_ranking = data.sort_values(
    "Our Sales",
    ascending=False
)

# Product ranking ka bar graph 
product_ranking.plot(
    x="Product",
    y="Our Sales",
    kind="bar",
    figsize=(10, 6),
    legend=False
)

# Graph ka title
plt.title("Product Performance Ranking")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Our Sales (Units)")

# Product names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Market mein hamari sales ka percentage calculate 
data["Market Share"] = (
    data["Our Sales"] / data["Market Sales"]
) * 100

# Product-wise market share ka bar graph 
data.plot(
    x="Product",
    y="Market Share",
    kind="bar",
    figsize=(10, 6),
    legend=False
)

# Graph ka title
plt.title("Product-wise Market Share")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Market Share (%)")

# Product names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Monthly sales ka sample data create 

monthly_data = pd.DataFrame({
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    
    "Our Sales": [
        5200,
        5600,
        6100,
        5900,
        6500,
        7000
    ]
})

# Monthly sales data display 
monthly_data
```

```python
# Monthly sales ka line graph 
monthly_data.plot(
    x="Month",
    y="Our Sales",
    kind="line",
    marker="o",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Monthly Sales Trend")

# X-axis ka naam
plt.xlabel("Month")

# Y-axis ka naam
plt.ylabel("Our Sales (Units)")

# Graph ko display 
plt.show()
```

```python
# Customer rating aur hamari sales ko compare 
data.plot(
    x="Product",
    y=["Customer Rating", "Our Sales"],
    kind="bar",
    figsize=(10, 6)
)

# Graph ka title
plt.title("Customer Rating vs Our Sales")

# X-axis ka naam
plt.xlabel("Product")

# Y-axis ka naam
plt.ylabel("Value")

# Product names ko seedha 
plt.xticks(rotation=0)

# Graph ko display 
plt.show()
```

```python
# Sabse zyada sales wala product find 
highest_sales_product = data.loc[
    data["Our Sales"].idxmax(),
    "Product"
]

# Sabse kam sales wala product find 
lowest_sales_product = data.loc[
    data["Our Sales"].idxmin(),
    "Product"
]

# Sabse bada sales gap wala product find 
highest_gap_product = data.loc[
    data["Sales Gap"].idxmax(),
    "Product"
]

# Sabse high rating wala product find 
highest_rating_product = data.loc[
    data["Customer Rating"].idxmax(),
    "Product"
]

# Sabse expensive product find 
highest_price_product = data.loc[
    data["Our Price"].idxmax(),
    "Product"
]

# Overall market share calculate 
overall_market_share = (
    data["Our Sales"].sum() /
    data["Market Sales"].sum()
) * 100

# Results display 
print("Highest Sales Product:", highest_sales_product)
print("Lowest Sales Product:", lowest_sales_product)
print("Highest Sales Gap Product:", highest_gap_product)
print("Highest Rated Product:", highest_rating_product)
print("Highest Priced Product:", highest_price_product)
print("Overall Market Share:", round(overall_market_share, 2), "%")
```
